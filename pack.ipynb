```python
import pandas as pd
import random

names = [
"Rahul Sharma","Priya Singh","Aman Verma","Neha Gupta","Arjun Yadav","Sneha Patel","Rohit Kumar","Pooja Mishra","Vikas Singh","Anjali Verma",
"Karan Gupta","Meena Yadav","Sachin Patel","Ritu Sharma","Deepak Kumar","Kavita Singh","Nikhil Verma","Shreya Gupta","Manoj Yadav","Aarti Patel",
"Abhishek Singh","Simran Kaur","Aditya Sharma","Riya Verma","Saurabh Gupta","Komal Yadav","Harsh Patel","Pallavi Mishra","Gaurav Singh","Nisha Kumari",
"Ankit Sharma","Divya Verma","Mohit Yadav","Sonal Gupta","Varun Singh","Payal Sharma","Akash Patel","Preeti Mishra","Rakesh Kumar","Sunita Devi",
"Vivek Singh","Anu Sharma","Rohini Gupta","Rajesh Yadav","Alok Verma","Pankaj Kumar","Seema Singh","Ravi Patel","Jyoti Sharma","Sandeep Yadav",
"Ashish Gupta","Rekha Verma","Mukesh Kumar","Anuradha Singh","Vinay Sharma","Shivani Gupta","Naveen Yadav","Sushma Devi","Rohit Verma","Priyanka Singh",
"Anand Kumar","Pooja Sharma","Ramesh Yadav","Neetu Singh","Lokesh Gupta","Amit Verma","Babita Sharma","Deepika Singh","Kunal Yadav","Mehul Patel",
"Sachin Kumar","Anita Verma","Ravindra Singh","Sarita Devi","Vishal Gupta","Priti Sharma","Dinesh Yadav","Kiran Kumari","Mahesh Verma","Sunil Sharma",
"Reena Gupta","Ajay Kumar","Savita Singh","Gopal Yadav","Nidhi Sharma","Tejas Patel","Bhavna Gupta","Hemant Verma","Shashi Kumar","Chandan Singh",
"Ritika Sharma","Yogesh Gupta","Kamlesh Yadav","Arti Verma","Narendra Singh","Geeta Devi","Sanjay Kumar","Madhu Sharma","Prakash Yadav","Lata Gupta"
]

data = []

for i in range(150):
    name = random.choice(names)
    attendance = random.randint(30, 95)
    quiz = round(random.uniform(2, 9), 1)
    assignment = random.randint(40, 95)
    im1 = random.randint(8, 25)
    im2 = random.randint(8, 25)
    prev = random.randint(50, 90)

    # Score calculation
    score = (attendance + quiz*10 + assignment + im1*2 + im2*2 + prev) / 6

    # Improved balanced logic
    if score < 50:
        result = 1   # clear fail
    elif score > 70:
        result = 0   # clear pass
    else:
        # STRONGER bias toward PASS (this fixes your issue)
        result = random.choices([0, 1], weights=[0.7, 0.3])[0]

    data.append([
        100+i, name, attendance, quiz, assignment,
        im1, im2, prev, result
    ])

df = pd.DataFrame(data, columns=[
    "student_id","student_name","attendance","quiz_avg",
    "assignment_rate","internal_marks_1","internal_marks_2",
    "previous_percentage","final_result"
])

df.to_csv("students.csv", index=False)

# Check balance
print("Dataset generated successfully!\n")
print(df["final_result"].value_counts())
```
